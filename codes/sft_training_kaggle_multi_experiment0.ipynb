## 3. Deney Listesi — EXPERIMENTS

- `"name"`         → çıktı klasörünün ve HF repo'sunun ismi (senin verdiğin isim)

- `"total_dataset_size"` → sadece `mix_mode="ratio"` iken kullanılır
- `"mix_config"`   → `(dosya_adı, miktar/oran)` çiftlerinin listesi

### Metodoloji — bu deneyler nasıl seçildi?

**1) Zayıf/güvenilmez datasetleri elendi.** Test accuracy'si raw baseline'ın
(%56) ALTINDA veya şu an güvenilmez olan datasetler mix'lere DAHİL EDİLMEDİ:
- Vanilla (%53 test) — baseline altı
- Farklı Yaklaşım 2k (%54 test) — 1k versiyonundan (%60) daha kötü, baseline altı
- Ön Bilgi 2k (%55 test) — baseline'a çok yakın/altı
- Kritik Nokta 1k (%36 test) — şu an yeniden değerlendiriliyor, sonucu belirsiz

**2) Kalan "iyi" datasetler, test accuracy'sine göre sıralandı:**

| Sıra | Dataset | Test acc | Kapasite |
|---|---|---|---|
| 1 | Tuzaklar 2k | %68 | max 2000 |
| 2 | Kritik Nokta 2k | %67 | max 2000 |
| 3 | Tuzaklar 1k | %63 | max 1000 (2k'nin alt kümesi) |
| 4 | Farklı Yaklaşım 1k | %60 | max 1000 |
| 5 | Ön Bilgi 1k | %57 | max 1000 |
| — | **Tüm Datasetler (5×1k stack)** | %68 | referans — kırılacak rekor |
| — | Raw baseline | %56 | — |

**3) İki farklı karışım stratejisi kullanılıyor:**
- **`disjoint` (2000 toplam)** → en iyi 2-3 dataset'i, TEK bir 2000'lik
  dataset'le (örn. "Tuzaklar 2k" tek başına %68) doğrudan aynı boyutta
  kıyaslamak için. Aynı GSM8K sorusu iki kez görünmez.
- **`stacked` (2000'i aşan, "Tüm Datasetler" tarzı)** → elindeki en iyi
  dataset'leri, her biri kendi kapasitesine kadar (max 2000) bağımsız
  olarak üst üste yığıp, "Tüm Datasetler" (%68, 5000 örnek, 5 dataset —
  bazıları zayıf) referansını SADECE en iyi 2-3 dataset'le, daha az
  toplam veriyle geçip geçemeyeceğini test etmek için.

## 4. Kaggle Dosyalarını Yükle (cache) & Karışım Fonksiyonu

Tüm deneylerde geçen Kaggle dosyalarını **bir kere** indirip belleğe alıyoruz
(aynı dosya birden fazla deneyde geçse bile tekrar indirmez).

```python
KAGGLE_DATASET = "sewowashere/soru-uzerinde-diyaloglar"
```

```python
_dataframes_cache = {}
```

```python
def get_dataframe(file_name):
    """Kaggle dosyasını cache'ten döndürür, yoksa indirir."""
    if file_name not in _dataframes_cache:
        print(f"Yükleniyor: {file_name}")
        df = kagglehub.load_dataset(
            KaggleDatasetAdapter.PANDAS,
            KAGGLE_DATASET,
            file_name,
        )
        df = df.reset_index(drop=True)
        print(f"  {len(df)} kayıt bulundu")
        _dataframes_cache[file_name] = df
    return _dataframes_cache[file_name]
```

```python
def _make_example(row):
    return {"messages": [
        {"role": "user", "content": str(row["question"])},
        {"role": "assistant", "content": str(row["model_output"])},
    ]}
```

```python
def build_mixed_dataset(mix_config, mix_mode="count", total_dataset_size=None, overlap_mode="disjoint"):
    """
    mix_config'teki (dosya_adi, miktar/oran) çiftlerinden birleşik bir dataset
    kurar. Seed KULLANILMAZ — her iki mod da tamamen deterministiktir.

    overlap_mode="disjoint" (varsayılan):
        Tüm dosyaların AYNI GSM8K soru havuzunu (aynı satır sırasıyla) paylaştığı
        varsayılır. Global, ÇAKIŞMAYAN bir soru-index'i üzerinden ardışık dilimler
        alınır — yani hiçbir soru iki dosyadan birden alınmaz. Tek dosyanın
        kapasitesiyle (elimizdeki max 2000) doğrudan karşılaştırılabilir mixler
        için kullan (örn. 2000'lik bir "tuzaklar 2k" ile aynı boyutta bir
        "tuzaklar+kritik-nokta" mixi).
        NOT: Dosyalar arasında kapasite farkı varsa (biri max 1000 diğeri max
        2000 gibi), düşük kapasiteli dosya otomatik olarak en düşük index
        aralığına yerleştirilir — elle sıralama uğraşmana gerek yok.

    overlap_mode="stacked":
        Her dosya KENDİ ilk N satırını BAĞIMSIZ olarak verir (her zaman 0'dan
        başlar). Böylece aynı GSM8K sorusu farklı öğretim tarzlarıyla birden
        fazla kez dataset'e girebilir — bu KASITLIDIR. "Tüm Datasetler" referans
        seti de böyle kurulmuştu (5 dosyadan 1000'er = 5000, her biri kendi ilk
        1000 sorusunu farklı bir çözüm tarzıyla veriyor). Tek bir dosyanın 2000
        satırlık kapasitesini AŞAN mixler kurmak istediğinde (örn. en iyi 2-3
        dataset'i "tüm datasetler" boyutuna yakın birleştirmek) bu modu kullan.

    Döner: (Dataset, mix_summary listesi) — mix_summary her zaman mix_config
    ile aynı sırada [(dosya, start, end, count), ...] döner.
    """

    def resolve_count(amount):
        if mix_mode == "ratio":
            if not total_dataset_size:
                raise ValueError("mix_mode='ratio' için total_dataset_size gerekli.")
            return round(amount * total_dataset_size)
        elif mix_mode == "count":
            return int(amount)
        else:
            raise ValueError(f"Bilinmeyen mix_mode: {mix_mode!r} ('count' ya da 'ratio' olmalı)")

    all_examples = []

    if overlap_mode == "disjoint":
        resolved = [(file_name, resolve_count(amount)) for file_name, amount in mix_config]
        # düşük kapasiteli dosyayı en düşük index aralığına otomatik yerleştir
        order = sorted(range(len(resolved)), key=lambda i: len(get_dataframe(resolved[i][0])))

        mix_summary = [None] * len(resolved)
        global_offset = 0
        for i in order:
            file_name, count = resolved[i]
            df = get_dataframe(file_name)
            start, end = global_offset, global_offset + count

            if end > len(df):
                raise ValueError(
                    f"'{file_name}' dosyası [{start}:{end}) aralığını karşılayamıyor "
                    f"(dosyada sadece {len(df)} kayıt var; disjoint modda otomatik "
                    f"sıralamayla bile sığmıyor). Toplam boyutu ya da miktarları azalt, "
                    f"ya da bu dataset kombinasyonu için overlap_mode='stacked' dene."
                )

            subset = df.iloc[start:end]
            all_examples.extend(_make_example(row) for _, row in subset.iterrows())
            mix_summary[i] = (file_name, start, end, count)
            global_offset = end

    elif overlap_mode == "stacked":
        mix_summary = []
        for file_name, amount in mix_config:
            count = resolve_count(amount)
            df = get_dataframe(file_name)
            if count > len(df):
                raise ValueError(
                    f"'{file_name}' dosyasında sadece {len(df)} kayıt var, "
                    f"{count} örnek istendi (stacked modda her dosya kendi [0:count) "
                    f"aralığını bağımsız verir)."
                )
            subset = df.iloc[0:count]
            all_examples.extend(_make_example(row) for _, row in subset.iterrows())
            mix_summary.append((file_name, 0, count, count))

    else:
        raise ValueError(f"Bilinmeyen overlap_mode: {overlap_mode!r} ('disjoint' ya da 'stacked' olmalı)")

    return Dataset.from_list(all_examples), mix_summary
```

```python
def print_mix_summary(mix_summary, overlap_mode="disjoint"):
    print(f"── Karışım özeti (overlap_mode={overlap_mode}) ──────────────────────")
    total = 0
    for file_name, start, end, count in mix_summary:
        print(f"  {file_name:<55} [{start:>5}:{end:<5})  {count:>5} örnek")
        total += count
    print(f"  {'TOPLAM':<55} {'':<14} {total:>5} örnek")
    print("──────────────────────────────────────────────────────────────")
```

## 6. Bildirim & Sanity-Check Yardımcı Fonksiyonları

```python
import requests
```

```python
def send_push_alert(title, message):
    api_key = "3193d93e-6601-4ea7-995f-f693a6a1139d"
    url = f"https://push.techulus.com/api/v1/notify/{api_key}"
    payload = {"title": title, "body": message}
    headers = {"Content-Type": "application/json"}
    try:
        response = requests.post(url, headers=headers, data=json.dumps(payload), timeout=10)
        if response.status_code == 200:
            print("Bildirim başarıyla gönderildi! ✅")
        else:
            print(f"Hata Kodu: {response.status_code} | Yanıt: {response.text}")
    except Exception as e:
        print(f"Bir bağlantı hatası oluştu: {e}")
```

```python
TEST_QUESTIONS = [
    "Ali'nin 5 elması, Ayşe'nin ise Ali'nin 3 katı kadar elması var. Toplam kaç elma vardır?",
    "Bir kitap 25 TL. Mert 4 kitap aldıysa ve 10 TL indirim aldıysa kaç TL ödedi?",
]
```

```python
def run_sanity_check(model, tokenizer):
    model.eval()
    for q in TEST_QUESTIONS:
        msgs = [
            {"role": "user", "content": f"Aşağıdaki matematik sorusunu adım adım çözer misin? Son cevabı '#### [Sayı]' formatında belirt.\n\nSoru: {q}"}
        ]
        prompt = tokenizer.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)
        inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
        with torch.no_grad():
            out = model.generate(
                **inputs,
                max_new_tokens=400,
                do_sample=False,
                pad_token_id=tokenizer.eos_token_id,
            )
        text = tokenizer.decode(out[0][inputs.input_ids.shape[1]:], skip_special_tokens=True)
        print(f"Q: {q}")
        print(f"A: {text[:500]}...\n{'-'*70}")
```

## 7. Ana Döngü — Tüm Deneyleri Sırayla Çalıştır

Her deney için: dataset kur → temiz model yükle → eğit → kaydet (Drive + yerel
+ opsiyonel HF push) → sanity-check → bildirim → belleği temizle → sıradaki deney.

```python
all_results = []
```

```python
for exp_idx, exp in enumerate(EXPERIMENTS, start=1):
    exp_name = exp["name"]
    mix_mode = exp.get("mix_mode", "count")
    total_dataset_size = exp.get("total_dataset_size")
    overlap_mode = exp.get("overlap_mode", "disjoint")
    mix_config = exp["mix_config"]

    print("\n" + "=" * 80)
    print(f"DENEY {exp_idx}/{len(EXPERIMENTS)}: {exp_name}")
    print("=" * 80)

    # ── 1) Dataset kur ────────────────────────────────────────────────────
    train_ds, mix_summary = build_mixed_dataset(mix_config, mix_mode, total_dataset_size, overlap_mode)
    print_mix_summary(mix_summary, overlap_mode)

    split = train_ds.train_test_split(test_size=0.05, seed=SEED)
    train_ds, eval_ds = split["train"], split["test"]
    print(f"Train: {len(train_ds)} | Eval: {len(eval_ds)}")

    # ── 2) Temiz model yükle ────────────────────────────────────────────
    model = load_fresh_model()
    print(f"Model yüklendi ({exp_name}). VRAM: {torch.cuda.memory_allocated()/1e9:.1f} GB")

    # ── 3) Trainer & eğitim ─────────────────────────────────────────────
    exp_output_dir = OUTPUT_DIR_BASE / exp_name
    exp_output_dir.mkdir(parents=True, exist_ok=True)

    sft_config = SFTConfig(
        output_dir=str(exp_output_dir),
        num_train_epochs=NUM_EPOCHS,
        per_device_train_batch_size=PER_DEVICE_BATCH_SIZE,
        per_device_eval_batch_size=PER_DEVICE_BATCH_SIZE,
        gradient_accumulation_steps=GRAD_ACCUM_STEPS,
        learning_rate=LEARNING_RATE,
        warmup_steps=50,
        weight_decay=WEIGHT_DECAY,
        lr_scheduler_type="cosine",
        optim="paged_adamw_8bit",
        bf16=(compute_dtype == torch.bfloat16),
        fp16=False,
        gradient_checkpointing=False,
        gradient_checkpointing_kwargs={"use_reentrant": False},
        max_length=MAX_SEQ_LENGTH,
        packing=False,
        assistant_only_loss=ASSISTANT_ONLY_LOSS,
        logging_steps=LOGGING_STEPS,
        save_steps=SAVE_STEPS,
        save_total_limit=2,
        eval_strategy="steps",
        eval_steps=EVAL_STEPS,
        report_to="none",
        seed=SEED,
        dataloader_num_workers=2,
        remove_unused_columns=True,
    )

    trainer = SFTTrainer(
        model=model,
        args=sft_config,
        train_dataset=train_ds,
        eval_dataset=eval_ds,
        processing_class=tokenizer,
    )

    train_result = trainer.train()
    final_loss = float(train_result.training_loss)
    print(f"[{exp_name}] Eğitim tamamlandı! Final loss: {final_loss:.4f}")

    # ── 4) Kaydet: Drive'daki SENİN VERDİĞİN isimdeki klasöre ──────────
    exp_save_dir = SAVE_DIR_BASE / exp_name
    trainer.save_model(str(exp_save_dir))
    tokenizer.save_pretrained(str(exp_save_dir))

    with open(exp_save_dir / "train_metrics.json", "w") as f:
        json.dump({
            "experiment_name": exp_name,
            "student": STUDENT_MODEL,
            "mix_mode": mix_mode,
            "overlap_mode": overlap_mode,
            "mix_config": mix_config,
            "mix_summary": mix_summary,
            "final_loss": final_loss,
            "epochs": NUM_EPOCHS,
            "train_size": len(train_ds),
            "eval_size": len(eval_ds),
            "timestamp": datetime.now().isoformat(),
        }, f, indent=2, ensure_ascii=False)

    print(f"[{exp_name}] Model kaydedildi: {exp_save_dir}")

    # ── 5) Sanity check ──────────────────────────────────────────────────
    run_sanity_check(model, tokenizer)

    # ── 6) Opsiyonel: HF Hub push ──────────────────────────────────────
    if PUSH_TO_HF:
        hf_repo = f"{HF_USERNAME}/qwen3_5-4b-{exp_name}"
        model.push_to_hub(hf_repo, token=HF_TOKEN, private=True)
        tokenizer.push_to_hub(hf_repo, token=HF_TOKEN, private=True)
        print(f"[{exp_name}] Push edildi: https://huggingface.co/{hf_repo}")

    # ── 7) Bildirim ───────────────────────────────────────────────────────
    send_push_alert(
        f"Deney bitti: {exp_name}",
        f"({exp_idx}/{len(EXPERIMENTS)}) final loss: {final_loss:.4f} — {exp_save_dir}",
    )

    all_results.append({
        "experiment_name": exp_name,
        "final_loss": final_loss,
        "train_size": len(train_ds),
        "eval_size": len(eval_ds),
        "save_dir": str(exp_save_dir),
    })

    # ── 8) Belleği temizle, sıradaki deneye hazırlan ────────────────────
    del trainer, model
    gc.collect()
    torch.cuda.empty_cache()
```
